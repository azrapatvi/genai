```python
import os
from dotenv import load_dotenv

load_dotenv()

os.environ['GROQ_API_KEY']=os.getenv("GROQ_API_KEY")

os.environ['LANGCHAIN_API_KEY']=os.getenv("LANGCHAIN_API_KEY")
```

```python
from langchain_groq import ChatGroq

llm = ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=0
)
```

the difference betwenn pydantic and type dict is after type dict wwe get the output as dictionary while with pydantic we get outpu as an object...
also pydantic do the data validation while type dict didnt do that...
with both of these we can get the structured output...

```python
from typing_extensions import TypedDict, Annotated

class Movie(TypedDict):
    name: Annotated[str, "The title or name of the movie"]
    year: Annotated[int, "The release year"]
    rating: Annotated[float, "The rating of the movie"]

model_with_structure = llm.with_structured_output(Movie)

response = model_with_structure.invoke(
    "Tell me about Inception movie. Give me the name, release year, and rating."
)

print(response)
```
