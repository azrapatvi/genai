```python
import os
from dotenv import load_dotenv

load_dotenv()

os.environ['GROQ_API_KEY']=os.getenv("GROQ_API_KEY")

os.environ['LANGCHAIN_API_KEY']=os.getenv("LANGCHAIN_API_KEY")
```

```python
from langchain_groq import ChatGroq

llm=ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0
)
```

```python
#now what is streaming?
# streaming means instead of getting the output from an llm once the aswer is generated fully ...what you can do is you can print teh output or nswer from the llm as it generatess....
# menas you dont need to wait for llm to completly gnerate the answer and thenn to get it...
# instead with srteaming you can get the llm response as soon as it generates 
# ex:your name is azra -> you will get this fully when llm generates without steaming
# ex: y
# o
# u 
# r 
# n 
# a 
# m 
# e 
# i 
# s 
# a 
# z 
# r 
# a 
# Streaming means getting the output from an LLM
# little by little as it is being generated,
# instead of waiting for the complete answer.

llm.invoke("what is genai")
```

```python
for chunk in llm.stream("what is genai?"):
    print(chunk.text,end=" ",flush=True) # flush =True means show the  ouptut on the screen immediatly ...menas as sson as it comes from llm show it directly on screen....without this sometimes ptyhon keeps the output in buffer  before displaying it so may not get output chunk by chunk
# you have see chat gpt how he reponsdss we get too see thee result come step by stepp ..little by littel
```

```python
# Batching = grouping multiple tasks/data items together and processing them as a batch.

responses=llm.batch([
    "What is GenAI?",
    "What is RAG?",
    "What is LangChain?"
])

for res in responses:
    print(res.content)
    print("***"*20)
    print()
```
