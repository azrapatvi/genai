```python
import os
from dotenv import load_dotenv

load_dotenv()

os.environ['GROQ_API_KEY']=os.getenv("GROQ_API_KEY")

os.environ['LANGCHAIN_API_KEY']=os.getenv("LANGCHAIN_API_KEY")
```

```python
from langchain_groq import ChatGroq

llm=ChatGroq(
    model='openai/gpt-oss-120b',
    temperature=0
)
```

pydantic is a librabyr which is used fro data validation...

```python
from pydantic import BaseModel,Field

# means i want output in this manner only...
class get_details(BaseModel):
    name:str=Field(description="name of the person")
    age:int=Field(description="age of the person")

```

```python
model_with_op_structure=llm.with_structured_output(get_details)  # this with_Structured_output is must; with it only we are defineing that we want this kind of structured output

model_with_op_structure
```

```python
model_with_op_structure.invoke("my name is azra patvi and i am 22 years old")
```

```python
# soo see we get the info the way we defined....
# so this then we can use to upload or insert in database...
```

```python
class movie(BaseModel):
    name:str=Field(description='the title or name of the movie')
    year:int=Field(description='the year when the movie was released')
    rating:float=Field(description='the ratings of the movie')
```

```python
movie_model_with_structure=llm.with_structured_output(movie)

movie_model_with_structure
```

```python
movie_model_with_structure.invoke("tell me about inception movie")
```

```python
# now we can aslo display the raw ai message 
class movie(BaseModel):
    name:str=Field(description='the title or name of the movie')
    year:int=Field(description='the year when the movie was released')
    rating:float=Field(description='the ratings of the movie')

movie_model_with_structure=llm.with_structured_output(movie,include_raw=True)
movie_model_with_structure.invoke("tell me about inception movie")
```

```python
#nested structure

class actor(BaseModel):
    name:str
    role:str

class movie_details(BaseModel):
    title:str=Field(description="the title of the movie")
    year:int=Field(description="the year when the movie released")
    actors:list[actor]=Field(description="Actors who worked in the movie") # list[actor] the actor is the 1st class we created and it is acctually list of actors
    genre:list[str]=Field(description="genre of the movie")
    budget:float|None=Field(description="the budget that required fot the movie")

movie_model_with_structure=llm.with_structured_output(movie_details)
movie_model_with_structure.invoke("privide details of movie inception")
```
