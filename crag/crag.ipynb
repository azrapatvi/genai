```python
#split the docuemnts

splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200
)

splitted_text=splitter.split_documents(loaded_data)

print("Data Splitted Successfully")
splitted_text
```

```python
print("total chunks: ",len(splitted_text))
```

```python
#create embeddings

embeddings=HuggingFaceEmbeddings(
    model_name='all-MiniLM-L6-v2'
)

```

```python
vector=FAISS.from_documents(
    embedding=embeddings,
    documents=splitted_text
)

#create retriever
retriever=vector.as_retriever(search_kwargs={"k":8})

print("FAISS database created successfully!")
```

```python
#document grader
# this will grade the documents based on releavance of the retrieved docs

grade_prompt=ChatPromptTemplate.from_template("""
    You are a document relevance evaluator.

    User Question:
    {question}

    Retrieved Document:
    {document}

    Check whether the retrieved document contains
    information useful for answering the question.

    Strictly Reply with ONLY one word:
    "YES" or "NO" only.
""")

grade_chain=grade_prompt|llm|StrOutputParser()
```

```python
grade_chain
```

```python
rewrite_prompt = ChatPromptTemplate.from_template("""

You are a search query optimizer.

Rewrite the following user question into a
better search engine query.

Make it clear and specific.

Return ONLY the improved search query.

Question:
{question}

""")

rewrite_chain=rewrite_prompt|llm|StrOutputParser()
```

```python
generation_prompt = ChatPromptTemplate.from_template("""

You are a helpful AI assistant.

Answer the question using ONLY the provided context.

If the answer is not available in the context,
clearly say that you don't have enough information togive answer based on the query.

Context:
{context}

Question:
{input}

Answer:

""")

document_chain=create_stuff_documents_chain(llm,generation_prompt)
```

```python
contextualize_prompt = ChatPromptTemplate.from_messages([
        ("system", """
        Given the chat history and latest user question,
        rewrite the question as a standalone question.
        Do not answer the question.
        """),
        MessagesPlaceholder("chat_history"),
        ("human", "{input}")
    ])

history_aware_retriever=create_history_aware_retriever(llm,retriever,contextualize_prompt)
```

```python
from langchain_core.chat_history import InMemoryChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory
```

```python
store = {}

def get_session_history(session_id: str):
    if session_id not in store:
        store[session_id] = InMemoryChatMessageHistory()

    return store[session_id]
```

```python
conversation_history=RunnableWithMessageHistory(
        rag_chain,
        get_session_history,
        input_messages_key='input',
        history_messages_key='chat_history',
        output_messages_key="answer"
    )
```

```python
def web_search(question):

    search=TavilySearchResults(max_results=5)
    results=search.invoke(question)

    web_documents = []

    for item in results:
        doc=Document(
            page_content=item.get('content'),
            metadata={
                'source':item.get('url')
            }
        )

        if doc.page_content.strip():
            web_documents.append(doc)

    return web_documents
    
```

```python
from langchain_core.messages import HumanMessage, AIMessage

chat_history = []   # FIX: simple memory
standalone_chain = contextualize_prompt | llm | StrOutputParser()   # FIX

def crag(question):
    print("CRAG")

    # FIX: turn follow-ups like "who is she" into a full question
    if chat_history:
        search_question = standalone_chain.invoke({
            "input": question,
            "chat_history": chat_history
        })
        print("Standalone question:", search_question)
    else:
        search_question = question

    #retriver
    docs = retriever.invoke(search_question)   # FIX: use the standalone question

    print("Retrieved documents:", len(docs))

    relevant_docs = []

    for doc in docs:
        res = grade_chain.invoke({'question': search_question, 'document': doc.page_content})   # FIX

        print("results are : ", res)
        result = res.strip().upper()

        print("Document grade:", result)

        if result.startswith("YES"):
            relevant_docs.append(doc)
            print("appended to relevant documents")

    if relevant_docs:

        print("\n. Relevant documents found!")
        print("Generating answer using PDF context...")

        # FIX: document_chain uses ONLY your graded docs
        # (rag_chain was ignoring relevant_docs and retrieving all 7 again)
        answer = document_chain.invoke({
            'input': search_question,
            'context': relevant_docs
        })

        source = "PDF"

    else:

        print("\n. No relevant documents found!")

        improved_query = rewrite_chain.invoke({"question": search_question})   # FIX
        print("Improved query:", improved_query)

        web_docs = web_search(improved_query)

        print("Web documents found:", len(web_docs))

        if web_docs:
            answer = document_chain.invoke({
                "input": search_question,
                "context": web_docs
            })   # FIX: removed config (it did nothing here)

        else:

            answer = "No relevant web search results found."

        source = "WEB"

    # FIX: save memory so follow-ups work
    chat_history.append(HumanMessage(content=question))
    chat_history.append(AIMessage(content=answer))

    return answer, source
```

```python

while True:

    question = input("\nAsk your question (type exit to stop): ")

    if question.lower().strip() == "exit":
        print("CRAG stopped.")
        break

    try:

        answer, source = crag(question)

        print("\n========== FINAL ANSWER ==========")

        print(answer)

        print("\nSOURCE:", source)

        print("==================================")

    except Exception as e:

        print("\nError:", str(e))
```
