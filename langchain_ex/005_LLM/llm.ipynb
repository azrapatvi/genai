```python
import os
from dotenv import load_dotenv

load_dotenv()

os.environ['GROQ_API_KEY']=os.getenv('GROQ_API_KEY')

os.environ['LANGCHAIN_API_KEY']=os.getenv("LANGCHAIN_API_KEY")
os.environ['LANGCHAIN_TRACING_V2'] = 'true'
os.environ['LANGCHAIN_PROJECT']=os.getenv("LANGCHAIN_PROJECT")
```

```python
# llm 
from langchain_groq import ChatGroq

llm=ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0
)
```

```python
llm
```

```python
from langchain_core.messages import HumanMessage,SystemMessage

messages=[
    SystemMessage(content="you are a translator; translate the given text from english to urdu."),
    HumanMessage(content="hello how are you?")
]

response=llm.invoke(messages)
```

```python
response
```

```python
response.content
```

```python
# also to get the response only from llm menas only the content then

from langchain_core.output_parsers import StrOutputParser

parser=StrOutputParser()

# so will directly pass the repsonse object becuase we already have the response from llm in there soo
parser.invoke(response)
```

```python
messages
```

```python
# using lcel we can chain the components
chain=llm|parser

chain.invoke(messages)
```

```python
#chat promopt tempalte

from langchain_core.prompts import ChatPromptTemplate

prompt=ChatPromptTemplate.from_messages([
    ("system","you are an expert language translator; translate the following into {language}"),
    ("user","{input}")
])

# now lets test teh prompt

prompt.invoke({"language":"Urdu","input":"hello how are you?"})
```

```python
# create chain

chain=prompt|llm|parser

response=chain.invoke({"language":"Urdu","input":"hello how are you?"})
```
