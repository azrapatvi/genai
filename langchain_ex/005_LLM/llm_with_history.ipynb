```python
# see the llm dont know you 
# so thats why wee need to maintain history
# now when i say history maintain it menas we have too provide our previous conversation as well to the model
# so that when we call the model model wiill be having previous chat history as well

# noww see 
# i will be giving the modell some chat history on own manually

from langchain_core.messages import HumanMessage,AIMessage

llm.invoke([
    (HumanMessage(content="hello my self azra patvi and i am a data scientist and an aiml engineer")),
    (AIMessage(content="Hello Azra! Nice to meet you and hear what you do. How can I assist you today?")),
    (HumanMessage(content="explain work of data scientist")),
    AIMessage(content="A data scientist gathers and cleans data, looks for patterns, and turns those insights into easy‑to‑understand recommendations for better decisions."),
    HumanMessage(content="do you remember me?who am i??")
])
```

```python
# so seee the abovee the model will now remember the things thatts what we doo when we say ading memory too llm

# message history
# we can use a message history to make the model stateful...without mesaage histroy the model is stateless
# this will keep track of input and outputs of the model response ...future interaction will then load those messages and pass them into llm so then llm remembers and knows the preivous chats as well

from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.runnables import RunnableWithMessageHistory

store={} # a dicitonary that will store the session id and there chat mssges

def get_session_history(session_id:str):
    if session_id not in store:
        store[session_id]=ChatMessageHistory()
    return store[session_id]


# runnable to add the historyy to the llm model
with_message_history=RunnableWithMessageHistory(llm,get_session_history)
```

```python
config1={"configurable":{"session_id":"chat1"}}

#name is of session_id is chat1 so if we change the session_id to hat2 then it will become new session 

res1=with_message_history.invoke(
    [
    HumanMessage(content="hello myself azra patvi and i am a data scientist and an aiml engineer")],
    config=config1
)

print(res1)
```

```python
res2=with_message_history.invoke(
    [HumanMessage(content=("what is my name?"))],
    config=config1
)
print(res2)
```

```python
config2={"configurable":{"session_id":"chat2"}}

res3=with_message_history.invoke(
    [HumanMessage(content="what is my name?")],
    config=config2
)
print(res3)
```

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

prompt=ChatPromptTemplate.from_messages([
    ("system","you are a helpful assistant, give all the naswers based on your ability"),
    MessagesPlaceholder(variable_name="messages"),
])

# create chain
chain= prompt | llm


with_message_history=RunnableWithMessageHistory(chain,get_session_history)
```

```python
config3={"configurable":{"session_id":"chat4"}}
with_message_history.invoke(
    {"messages":[HumanMessage(content="do you know my name?")]},
    config=config3
)
```

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

prompt=ChatPromptTemplate.from_messages([
    ("system","you are a helpful assistant, give all the naswers based on your ability in this language: {language}"),
    MessagesPlaceholder(variable_name="messages"),
])

# create chain
chain= prompt | llm

with_message_history=RunnableWithMessageHistory(chain,get_session_history,input_messages_key="messages")
```

```python
# managing the conversation hsitory
# why ?
# because see if we dont mnagae the history the history size will increase and will become unknowingly so much big in such that it will cross the modles context windoww
# therefor ti is impto limit the chat history from growing unnecessaritly

from langchain_core.messages import SystemMessage,trim_messages

trimmer=trim_messages(
    max_tokens=70, # will keep exactly 70 tokens or words 
    strategy='last', # if the lenght of your text is more than 70 tokens then it will trim the starting messages andwill exatly kepp 70 tokens
    token_counter=llm,
    include_system=True, #include system messgaes
    allow_partial=False,
    start_on='human'
)
```

```python
from operator import itemgetter
from langchain_core.runnables import RunnablePassthrough # thsi si just too pass the inptu forward ; just take he inptu and pass it forward


chain = (
    RunnablePassthrough.assign(
        messages=itemgetter("messages") | trimmer
    )
    | prompt
    | llm
)

```
