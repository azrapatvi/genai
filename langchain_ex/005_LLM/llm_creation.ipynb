```python
#prompt template is like how you want your model to behavee ; 
# measn givng prompt to ai model to pbehave like thiss and thiss you are friendly ai agent

# chatprompt templte

from langchain_core.prompts import ChatPromptTemplate

prompt=ChatPromptTemplate.from_messages(
    [
        ("system","you are a very friendly ai assistant!"),
        ("user","{input}")
    ]
)

prompt
```

```python
# create a chain betwenn llm and prompt

chain=prompt | llm

response=chain.invoke({"input":"who are you??"}) # use input variable here only becuase the place holder we define in the prompt is {input} so thats when invoking it use {input}
print(response)
```

```python
response.content
```

```python
# output parser means how you want the output from the llm 
# response in specifc format like json and all

from langchain_core.output_parsers import StrOutputParser

parser=StrOutputParser()

chain2=prompt|llm|parser

res2=chain2.invoke({"input":"who are you?"})
print(res2)
```
