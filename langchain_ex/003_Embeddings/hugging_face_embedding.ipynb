```python
from langchain_huggingface import HuggingFaceEmbeddings

embeddings=HuggingFaceEmbeddings(
    model_name="all-MiniLM-L6-v2"
)
```

```python
text="my self azra patvi"

vector=embeddings.embed_query(text)

vector
```

```python
print("Embedding dimensions:", len(vector))
```

```python
documents = [
    "RAG stands for Retrieval Augmented Generation.",
    "RAG retrieves relevant information from a knowledge base.",
    "Embeddings convert text into numerical vectors."
]

vectors_of_doc=embeddings.embed_documents(documents)


vectors_of_doc
```

```python
print("Embedding dimensions:", len(vectors_of_doc)) # len will be 3 becuae we have 3 sentences
```
