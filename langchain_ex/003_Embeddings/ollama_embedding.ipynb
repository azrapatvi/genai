##### Ollama Embeddings

```python
from langchain_community.embeddings import OllamaEmbeddings

embeddings=OllamaEmbeddings(
    model='nomic-embed-text'
)
```

```python
embeddings
```

```python
text = "Generative AI creates new content."

vector=embeddings.embed_query(text) # here the text i have is onyl a string thast why

print(vector[:10])
print(len(vector))
```

```python
vectors=embeddings.embed_documents(
    ["Generative AI creates new content.",
     "RAG uses external documents to answer questions."]
)

print(len(vectors))
print(len(vectors[0]))
print(len(vectors[1]))
```
