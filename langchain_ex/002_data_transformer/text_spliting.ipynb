##### Spliting the text from document

```python
# reading a pdf file

from langchain_community.document_loaders import PyPDFLoader

loader=PyPDFLoader(r"ugc_ad_script.pdf")
```

```python
loader.load()
```

```python
docs=loader.load()
```

```python
docs
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter=RecursiveCharacterTextSplitter(chunk_size=500,chunk_overlap=50) # it menas in each chunks or groups i want 500 total characters and chunk overlap meanss i want 50 charcaters overlap in new chuck ex chunk1 :abc...xyz.50chars chunk2 :xyz.50chars.lmnfdfd
```

```python
text_splitter
```

```python
final_doc=text_splitter.split_documents(docs)

final_doc
```

```python
print(final_doc[0])
```

```python
print(final_doc[1])
```

```python
# nwo similaryl lets split the text data from txt file

from langchain_community.document_loaders import TextLoader

loader=TextLoader("speech.txt")
docs=loader.load()
```

```python
docs
```

```python
text_splitter=RecursiveCharacterTextSplitter(chunk_size=100,chunk_overlap=20)

splitted_text = text_splitter.split_documents(docs)

splitted_text
```

```python
splitted_text[0]
```

```python
splitted_text[1]
```

```python
text="coffee is one of the most popular drinks in the world. Every day, people drink more than two billion cups of it. But the story of coffee began long ago, in the highlands of Ethiopia, and it is full of surprising turns."
text_splitter2=RecursiveCharacterTextSplitter(chunk_size=50,chunk_overlap=15)
splitted_text2=text_splitter2.create_documents([text]) # use create documents when the text you have is in the list type and use split documents when the data you have loaded from textlaoder or any loader
splitted_text2
```
