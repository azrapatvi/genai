```python
from langchain_text_splitters import CharacterTextSplitter

text = """Coffee is one of the most popular drinks in the world.

Every day, people drink more than two billion cups of it.

The story of coffee began long ago in Ethiopia."""

# with this character text splitter we can decide the split menas from where we want to split...
# character textsplitter is basically when we want to split by a specific seprator

text_splitter=CharacterTextSplitter(
    separator="\n\n",
    chunk_size=50,
    chunk_overlap=15
)

splitted_text=text_splitter.create_documents([text])
splitted_text
```

```python
from langchain_text_splitters import CharacterTextSplitter

text = """Coffee is one of the most popular drinks in the world.

Every day, people drink more than two billion cups of it.

The story of coffee began long ago in Ethiopia."""


text_splitter=CharacterTextSplitter(
    separator=" ",
    chunk_size=50,
    chunk_overlap=15
)

splitted_text=text_splitter.create_documents([text])
splitted_text
```

now notice the differnce here in the above where the seprator is space so it is considering the spaces for split but when the seprator was \n\n so it was considering that \n\n....so thats the difference between RecursiveCharacterTextSplitter and ChracterTextSplitter...
