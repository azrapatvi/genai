```python
from langchain_groq import ChatGroq
from langchain.tools import tool
from langchain.agents import create_agent
import os
from dotenv import load_dotenv
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain.messages import HumanMessage,AIMessage,SystemMessage

load_dotenv()

os.environ['GROQ_API_KEY']=os.getenv('GROQ_API_KEY')
```

```python
llm=ChatGroq(model='openai/gpt-oss-120b')
```

```python
short_prompt=ChatPromptTemplate.from_messages([
    "explain {topic} in short."
])
long_prompt=ChatPromptTemplate.from_messages([
    "explain {topic} in detail."
])

```

```python
from langchain_core.output_parsers import StrOutputParser

parser=StrOutputParser()
```

```python
from langchain_core.runnables import RunnableParallel
chains={
    "short":short_prompt|llm|parser,
    "detail":long_prompt|llm|parser
}
# the dictinary is not a runnable so we cant access it using invoke thats why

chains=RunnableParallel(chains)

res=chains.invoke({'topic':'machine learning'})

print(res)
```

```python
# now i want to give different topics to these chains

from langchain_core.runnables import RunnableParallel,RunnableLambda
chains={
    "short":RunnableLambda(lambda x :x['short'])|short_prompt|llm|parser,
    "detail":RunnableLambda(lambda x :x['detail'])|long_prompt|llm|parser
}
# the dictinary is not a runnable so we cant access it using invoke thats why

chains=RunnableParallel(chains)

res=chains.invoke({
    'short':{'topic':'machine learning'},
    'detail':{'topic':'deep learning'}
})

print(res)
```
